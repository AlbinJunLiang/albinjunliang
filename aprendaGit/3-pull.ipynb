#### 3 - What do i have to do if someone make a change in a remote repository and i want to get the changes?

1. if you want to do it, you should have cloned the repository beforehand.
2. Execute the next command:

```python
git pull

# Specific branch local to update
git pull origin <BRANCH_NAME>



# If you only need get changes from the remote repository
git config pull.rebase false  
```
