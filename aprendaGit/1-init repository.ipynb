#### 1 - If a initialize a repository locally before creating it on github, what should i do?

1. Go to github and create a new repository
2. Do not check README, .gitignore or license if you already have files locally.
3. Configure git locally in your project



```python
git add .
git commit -m "Initial commit"
```

4. Connect your local repository to the remote repository

```python
git branch -M main
git remote add origin https://github.com/USUARIO/REPOSITORIO.git
git push -u origin main
```

5. Upload the files to your remore repository

```python
git add .
git commit -m "Descripción de los cambios"
git push
```
