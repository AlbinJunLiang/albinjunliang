#### Function to move accents

```python

def quitar_tildes(texto, move = True):
    if move:
                # Asegura que sea texto y normaliza a formato NFD
        texto_normalizado = unicodedata.normalize('NFD', str(texto))
        # Filtra y descarta los caracteres de acentuación (Category 'Mn')
        texto_sin_tildes = "".join(c for c in texto_normalizado if unicodedata.category(c) != 'Mn')
        return texto_sin_tildes
    else:
        return texto
```

```python
# Obtener los nombres de las columnas en una lista
columns = df.columns.tolist()
print(columns)



columns = [quitar_tildes(i, True) for i in columns]
```

### Get only the data values

```python
data = df.values.tolist()
print(data)
```

### CSV FIle to SQL

```python
print(len(data))

for i in data:
    for j in range(len(i)):
        print(i[j], end=" ")
    print("\n")

```

```python
output = f"INSERT INTO {file_name} ({','.join(columns)}) \nVALUES "

filas = []

for i in data:
    # Si el dato es texto (str), le pone comillas ''. Si es número, lo deja igual.
    fila_formateada = "(" + ",".join(f"'{j}'" if isinstance(j, str) else str(j) for j in i) + ")"
    filas.append(fila_formateada)

# Unes todas las filas y las sumas al output final
values = ",\n".join(filas)
final_query = output + values + ";"

print(final_query)
```
